```python
import os
import sys
import pandas as pd
from src.utils.db_utils import get_connection, execute_query

def playertabel():
    """Player table with playerid"""
    query = """
    Select *
    from stats.player
    where playerid = 'HopkDe00'
    """
    return  execute_query(query)
player_table = playertabel()

print("player Table")
print(player_table)

def playerhistory():
    """Playerhistory table with playerid"""
    query = """
    select playerid, playername, season, teamid, position, games_played, games_started, trade_week, total_snaps, total_off_snaps, avg_off_pct
    from stats.playerhistory
    where playerid = 'HopkDe00'
    order by season desc;


    """
    return  execute_query(query)
player_table_history = playerhistory()

print(" ")
print("Player History")
print(player_table_history)

def playerstat():
    """Playerhistory table with playerid"""
    query = """
    SELECT playerid, player, season, week, teamid, targets, rec, rec_yds, rec_yac, rec_td
    FROM stats.receiving
    WHERE playerid = 'HopkDe00'
    ORDER BY season DESC, week ASC;


    """
    return  execute_query(query)
player_stat = playerstat()

print(" ")
print("Player Stats")
print(player_stat)

```
