## Step 2: View Available Tables
Run this cell to see what tables are available in the database:

```python
# Query to list all tables in the public schema
query = """
SELECT table_name, table_type
FROM information_schema.tables 
WHERE table_schema = 'stats'
ORDER BY table_name;
"""

tables_df = execute_query(query)
print('Tables in database:')
display(tables_df)
```
