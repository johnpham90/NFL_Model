# Test Supabase Database Connection

This notebook tests the connection to our Supabase database and performs some basic queries.

```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Get database URL and modify it to use postgresql:// instead of postgres://
DATABASE_URL = os.getenv('DATABASE_URL').replace('postgres://', 'postgresql://')

# Create SQLAlchemy engine
engine = create_engine(DATABASE_URL)

# Test connection
print('Connection successful!')
```

## Test Query
Let's try a simple query to list all tables in our database

```python
# Query to get all tables
query = """
SELECT *
FROM stats.gamesummary
WHERE season = '2024';
"""

# Execute query and load results into DataFrame
tables_df = pd.read_sql(query, engine)
print('\nTables in database:')
display(tables_df)
```
